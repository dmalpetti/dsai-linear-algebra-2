# Eigenspaces of orthogonal projection and symmetry on a line

```python
import numpy as np
from sympy import Matrix, Rational, eye, pprint

from plot3d_utils import new_figure, draw_axes, draw_plane, draw_line, draw_vector
```

We consider the line $r$ through the origin with direction $\mathbf{d}=(1,1,1)$, namely
$$r=\langle(1,1,1)\rangle.$$

Using the formula
$$P=\frac{\mathbf{d}\,\mathbf{d}^T}{\|\mathbf{d}\|^2},$$
we can calculate the matrix of the orthogonal projection onto $r$: this time we keep the component along $\mathbf{d}$, instead of removing it.

```python
d = Matrix([1, 1, 1])
P = (d @ d.T) / d.norm()**2
pprint(P)
```

Using the analogous formula
$$S=2\,\frac{\mathbf{d}\,\mathbf{d}^T}{\|\mathbf{d}\|^2}-I,$$
we can also calculate the matrix of the symmetry with respect to $r$: the component along $\mathbf{d}$ is kept, while the orthogonal one is reversed.

```python
S = 2 * (d @ d.T) / d.norm()**2 - eye(3)
pprint(S)
```

The functions that draw the axes, the line and the vectors are in `plot3d_utils.py`, imported above. With these elements we can explore the effect of the matrix on vectors and then deduce the eigenvalues and eigenspaces.

Besides the line $r$, we also draw the plane through the origin orthogonal to it: it is the other eigenspace.

We start by analyzing the projection.

```python
v = Matrix([1, 1, -])
pv = P @ v

fig = new_figure("Orthogonal projection onto the line r")
draw_axes(fig)
draw_plane(fig, d, name="plane orthogonal to r")
draw_line(fig, d, "black", "line r")
draw_vector(fig, v, "orange", "<b>v</b>")
draw_vector(fig, pv, "red", "P <b>v</b>")

fig
```

We now do the same with the symmetry.

```python
# the same vector, reflected with respect to the line
sv = S @ v

fig = new_figure("Symmetry with respect to the line r")
draw_axes(fig)
draw_plane(fig, d, name="plane orthogonal to r")
draw_line(fig, d, "black", "line r")
draw_vector(fig, v, "orange", "<b>v</b>")
draw_vector(fig, sv, "red", "S <b>v</b>")
fig
```
