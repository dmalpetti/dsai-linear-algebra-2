# Eigenspaces of orthogonal projection and symmetry on a line in $\mathbb{R}^2$

```python
import numpy as np
from sympy import Matrix, Rational, eye, pprint

from plot2d_utils import new_figure, draw_axes, draw_line, draw_vector
```

We consider the line $r$ through the origin with direction $\mathbf{u}=(3,-1)$, namely
$$r=\langle(3,-1)\rangle.$$

Using the formula
$$P=\frac{\mathbf{u}\,\mathbf{u}^T}{\|\mathbf{u}\|^2},$$
we can calculate the matrix of the orthogonal projection onto $r$: of a vector, we keep only the component along $\mathbf{u}$.

```python
u = Matrix([3, -1])
P = (u @ u.T) / u.norm()**2
pprint(P)
```

Using the analogous formula
$$S=2\,\frac{\mathbf{u}\,\mathbf{u}^T}{\|\mathbf{u}\|^2}-I,$$
we can also calculate the matrix of the symmetry with respect to $r$: the component along $\mathbf{u}$ is kept, while the orthogonal one is reversed.

```python
S = 2 * (u @ u.T) / u.norm()**2 - eye(2)
pprint(S)
```

The functions that draw the axes, the lines and the vectors are in `plot2d_utils.py`, imported above. With these elements we can explore the effect of the matrix on vectors and then deduce the eigenvalues and eigenspaces.

Besides the line $r$, we also draw the line orthogonal to it, $\langle(-1,2)\rangle$: it is the other eigenspace.

We start by analyzing the projection.

```python
v = Matrix([1, 3])
pv = P @ v

fig = new_figure("Orthogonal projection onto the line r")
draw_axes(fig)
draw_line(fig, u, "black", "line r")
draw_line(fig, Matrix([1, 3]), "light blue", "line orthogonal to r")
draw_vector(fig, v, "green", "<b>v</b>")
draw_vector(fig, pv, "light green", "P <b>v</b>")

fig
```

We now do the same with the symmetry.

```python
# the same vector, reflected with respect to the line
sv = S @ v

fig = new_figure("Symmetry with respect to the line r")
draw_axes(fig)
draw_line(fig, u, "black", "line r")
draw_line(fig, Matrix([-1, 2]), "light blue", "line orthogonal to r")
draw_vector(fig, v, "orange", "<b>v</b>")
draw_vector(fig, sv, "red", "S <b>v</b>")
fig
```
