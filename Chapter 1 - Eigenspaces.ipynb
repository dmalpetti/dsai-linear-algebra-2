# Exploration of eigenspaces

```python
from sympy import Matrix, pprint

from plot3d_utils import new_figure, draw_axes, draw_plane, draw_line, draw_vector
```

```python
A = Matrix([[1, 1, 1], [-1, 3, 1], [-1, 1, 3]])
pprint(A)
```

```python
# eigenvalues and associated eigenvectors
# returns: (eigenvalue, algebraic_multiplicity, [eigenvectors])
Aeigvecs = A.eigenvects()
pprint(Aeigvecs)
```

The functions that draw the axes and the vectors are in `plot3d_utils.py`, imported above. With these elements we can explore the effect of the matrix on vectors and then comment on the eigenvalues and eigenspaces.

```python
v1 = Aeigvecs[0][2][0]  # eigenvector associated with eigenvalue 2
#pprint(v1)
v2 = Aeigvecs[0][2][1]  # eigenvector associated with eigenvalue 2
v3 = Aeigvecs[1][2][0]  # eigenvector associated with eigenvalue 3

fig = new_figure("Exploration of Eigenspaces")
draw_axes(fig)
draw_vector(fig, v1, "orange", "<b>v1</b>")
draw_vector(fig, A@v1, "light orange", "<b>Av1</b>")
draw_vector(fig, v2, "red", "<b>v2</b>")
draw_vector(fig, A@v2, "light red", "A<b>v2</b>")
draw_vector(fig, v3, "green", "<b>v3</b>")
draw_vector(fig, A@v3, "light green", "A<b>v3</b>")

fig
```

The function below generates some vectors of the span of a list of vectors: each one is a linear combination of the given vectors, with coefficients chosen at random between -1 and 1.

```python
import numpy as np

def generate_span(vectors, n=10, low=-1, high=1):
    """n random linear combinations of the given vectors, with coefficients between low and high."""
    combinations = []
    for _ in range(n):
        coefficients = np.round(np.random.uniform(low, high, len(vectors)), 2)
        combination = sum((c * v for c, v in zip(coefficients, vectors)), Matrix.zeros(*vectors[0].shape))
        combinations.append(combination.applyfunc(lambda x: round(x, 2)))   # shows 0.37, not 0.370000000000000
    return combinations


span = generate_span([v1,v2], n=20)
#pprint(span)

# for v in span:
#     draw_vector(fig, v, "blue", "<b>v</b>", legend=False)
#     draw_vector(fig, A@v, "light blue", "A<b>v</b>", legend=False)

# draw_plane(fig, [v1, v2], name="plane α")
```

```python
fig = new_figure("Exploration of Eigenspaces")
draw_axes(fig)
draw_vector(fig, v1, "orange", "<b>v1</b>")
draw_vector(fig, A@v1, "light orange", "<b>Av1</b>")
draw_vector(fig, v2, "red", "<b>v2</b>")
draw_vector(fig, A@v2, "light red", "A<b>v2</b>")
draw_vector(fig, v3, "green", "<b>v3</b>")
draw_vector(fig, A@v3, "light green", "A<b>v3</b>")

for v in span:
    draw_vector(fig, v, "blue", "<b>v</b>", legend=False)
    draw_vector(fig, A@v, "light blue", "A<b>v</b>", legend=False)

draw_plane(fig, [v1, v2], name="plane α")

fig
```
