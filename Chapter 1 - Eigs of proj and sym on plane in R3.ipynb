# Eigenspaces of orthogonal projection and symmetry on a plane

```python
import numpy as np
from sympy import Matrix, Rational, eye, pprint

from plot3d_utils import new_figure, draw_axes, draw_plane, draw_line, draw_vector
```

We consider the plane $\alpha$ through the origin with normal $\mathbf{n}=(1,-2,2)$, namely
$$\alpha:\ x-2y+2z=0.$$

Using the formula
$$P=I-\frac{\mathbf{n}\,\mathbf{n}^T}{\|\mathbf{n}\|^2},$$
we can calculate the matrix of the orthogonal projection onto $\alpha$.

```python
n = Matrix([1, -2, 2])
P = eye(3) - (n @ n.T) / n.norm()**2
pprint(P)
```

Using the analogous formula
$$S=I-2\,\frac{\mathbf{n}\,\mathbf{n}^T}{\|\mathbf{n}\|^2},$$
we can also calculate the matrix of the symmetry with respect to $\alpha$: instead of removing the component along $\mathbf{n}$, we reverse it.

```python
S = eye(3) - 2 * (n @ n.T) / n.norm()**2
pprint(S)
```

The functions that draw the axes, the plane and the vectors are in `plot3d_utils.py`, imported above. With these elements we can explore the effect of the matrix on vectors and then deduce the eigenvalues and eigenspaces.

We start by analyzing the projection.

```python
v = Matrix([2, 1, 1])
pv = P @ v

fig = new_figure("Orthogonal projection onto the plane α")
draw_axes(fig)
draw_plane(fig, n, name="plane α")
draw_line(fig, n, "black", "line n")
draw_vector(fig, v, "green", "<b>v</b>")
draw_vector(fig, pv, "light green", "P <b>v</b>")

fig
```

We now do the same with the symmetry.

```python
# the same vector, reflected with respect to the plane
sv = S @ v

fig = new_figure("Symmetry with respect to the plane α")
draw_axes(fig)
draw_plane(fig, n, name="plane α")
draw_line(fig, n, "black", "line n")
draw_vector(fig, v, "green", "<b>v</b>")
draw_vector(fig, sv, "light green", "S <b>v</b>")
fig
```
